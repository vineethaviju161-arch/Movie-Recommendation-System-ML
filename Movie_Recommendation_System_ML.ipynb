# 🎬 Movie Recommendation System (Content-Based)

🔹 Project Goal

Recommend similar movies based on a movie selected by the user.

Example:

Input:
Batman Begins

Output:

The Dark Knight
Man of Steel
Inception

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Load dataset
movies = pd.read_csv("C:\DATASCIENCE\MYPROJECTS\movies.csv")
```

```python
# Convert text into features
cv = CountVectorizer()
vectors = cv.fit_transform(movies["genre"]).toarray()
```

```python
# Similarity matrix
similarity = cosine_similarity(vectors)
```

```python
# Recommendation function
def recommend(movie_name):
    
    index = movies[movies["movie"] == movie_name].index[0]
    
    distances = list(enumerate(similarity[index]))
    
    recommended = sorted(
        distances,
        reverse=True,
        key=lambda x:x[1]
    )[1:4]
    
    print("Recommended Movies:")
    
    for i in recommended:
        print(movies.iloc[i[0]].movie)
```

```python
# Test
recommend("Batman Begins")
```
